## 3. AI 수학 기초

신경망의 계산은 선형 관계식에서 시작함

$$y = w \cdot x + b$$

1. w (weight, 가중치) : 입력에 곱해지는 값
2. b (bias, 편향) : 계산 결과에 더해지는 값

학습은 이 w와 b를 데이터에 맞는 값으로 찾아가는 과정임

### 3-1. 학습의 과정

**학습(training)이란**
- 한 번의 계산으로 답을 구하는 것이 아니라, 오차를 줄이는 방향으로 w, b를 여러 번 고쳐 나가는 과정

**학습 순서**
1. w, b를 무작위로 초기화함 → 처음 예측은 틀림
2. 한 번 계산해서 오차를 구함
3. 오차를 줄이는 방향으로 w, b를 조금씩 수정함
4. [예측 → 오차 계산 → w, b 수정]을 수백~수천 번(epoch) 반복함
   - 반복할수록 오차가 줄고 w, b가 정답에 가까워짐

이 반복에서 "예측"이 순전파, "w, b를 수정"하는 쪽이 역전파임 (§3-3)

### 3-2. 최적의 weight와 경사 하강법

**최적의 weight란**
- 오차(loss)가 가장 작아지는 w, b 값
- 오차가 0에 가까울수록 예측이 실제값에 가까움

**경사 하강법 (Gradient Descent)**
- 오차가 최소인 지점을 찾아가는 방법
- 오차 함수의 기울기(경사)를 따라 내려가면서 w를 갱신함
- 기울기가 0에 가까워지는 지점이 오차가 최소인 지점(최적의 weight)임

### 3-3. 순전파(Forward)와 역전파(Backward)

각 w, b의 기울기를 실제로 구하는 방법이 역전파임

**종류**
1. 순전파 (forward propagation) : 입력 → 출력 방향
   - x를 층에 차례로 통과시켜 예측값 ŷ를 계산함
2. 역전파 (backpropagation) : 출력 → 입력 방향
   - loss를 거꾸로 전달하며 각 가중치의 기울기를 계산함

**학습 1스텝의 흐름**

```
1. 순전파    x → 층들 통과 → 예측값 ŷ
2. 손실 계산  loss = mse(y, ŷ)
3. 역전파    loss를 거꾸로 전달해 각 w, b의 기울기(gradient)를 구함
4. 갱신      optimizer(adam)가 기울기만큼 w, b를 실제로 수정
```

- 추론은 순전파만 함 (1번)
- 학습은 순전파 + 역전파 + 갱신을 모두 함 (1~4번)
- 역전파는 기울기를 구하는 계산이고, 가중치를 실제로 바꾸는 것은 4번의 optimizer임

## 4. AI·ML·DL·LLM의 관계

![AI, ML, DL, LLM의 포함 관계](assets/ai-ml-dl-llm.svg)

**포함 관계**
1. AI (인공지능) : 가장 넓은 개념
2. ML (머신러닝) : AI의 한 종류
3. DL (딥러닝) : 머신러닝의 한 종류
4. LLM (거대 언어모델) : 딥러닝의 한 종류
   - Layer를 아주 깊고 크게(파라미터 방대) 쌓아 언어 데이터를 학습시킨 딥러닝 모델
   - 구조 원리는 같고, 층 수·노드 수·데이터 양의 규모만 훨씬 큼

### 4-1. 머신러닝과 딥러닝의 공통 원리

1. 예측값을 만듦 (`y = w*x + b` 같은 함수)
2. 실제값과 예측값의 차이(오차, loss)를 계산함
3. 오차를 줄이는 방향으로 w, b를 조금씩 수정함 (경사 하강법)
4. 이 과정을 반복해서 최적의 w, b를 찾음

### 4-2. 머신러닝과 딥러닝의 차이

| 구분 | 구조 | 특징 |
|---|---|---|
| **머신러닝** (전통 방식) | `y = w*x + b` 계산을 단순한 구조로 1~소수 개만 사용 (선형회귀, 로지스틱회귀, 결정트리 등) | 사람이 어떤 feature를 볼지 미리 설계해야 하는 경우가 많음 |
| **딥러닝** (Deep Learning) | Node → Layer → 여러 Layer를 깊게(Deep) 쌓아올린 구조 | 층이 깊어질수록 더 복잡한 패턴을 스스로 찾아냄 |

- 딥러닝은 여러 층(Layer)을 깊게 쌓은 머신러닝임
- 경사 하강법으로 w, b를 찾는다는 학습 원리는 같음

## 5. 인공지능 코딩의 4단계

1. 데이터 준비 : x, y
2. 모델 구성 : `Sequential` + `Dense`
3. 컴파일 & 훈련 : `compile`, `fit`
4. 평가 & 예측 : `evaluate`, `predict`

### 5-1. compile(loss, optimizer)

**구성 요소**
1. loss : 오차를 측정하는 기준
   - `mse` (Mean Squared Error, 평균제곱오차) : 오차를 제곱해서 평균 낸 값
   - `mae` (Mean Absolute Error, 평균절대오차) : 오차의 절댓값을 평균 낸 값
2. optimizer : loss를 최소화하는 방향으로 w, b를 실제로 갱신하는 알고리즘 (경사 하강법 계열)
   - 대표 : `adam`, `sgd`, `rmsprop`

**둘의 역할 구분**
- loss는 얼마나 틀렸는지를 재는 기준
- optimizer는 그 오차를 보고 w, b를 어떻게 고칠지 정함

### 5-2. fit / evaluate / predict

| 메서드 | 역할 | 내부 동작 |
|---|---|---|
| `model.fit(x, y, epochs=n)` | epoch 수만큼 [예측 → 오차 계산 → w,b 수정]을 반복하며 학습 | 순전파 + 역전파 + 갱신 |
| `model.evaluate(x, y)` | compile에서 지정한 loss 방식으로 지금 모델 상태의 오차를 계산 | 순전파 + 손실 계산 (역전파 없음) |
| `model.predict(x)` | 마지막으로 학습된 w, b를 대입해서 값을 계산 | 순전파만 |

**규칙**
1. `evaluate`는 loss를 출력하지만 역전파를 하지 않아 가중치가 변하지 않음 (학습이 아님)
2. `predict`가 `fit`보다 빠른 이유도 순전파 한 단계만 하기 때문임
3. epoch를 많이 돌린다고 항상 좋은 것은 아님
   - 어느 지점부터 loss가 정체되거나 과적합이 생길 수 있음
4. 데이터가 완벽한 비례 관계가 아니면 loss가 0으로 수렴하지 못함
   - 예: x=[1,2,3,4,5], y=[1,2,4,3,5]는 모든 점을 지나는 직선이 없음
   - 오차가 가장 작은 최적의 직선에서 수렴함

## 6. 신경망의 구성 요소 - Node / Layer / Sequential

신경망은 `y = w*x + b` 계산을 여러 층(layer)으로 쌓고, 각 층마다 여러 번(node) 수행하는 구조임

**구성 요소**
1. Node (=Unit, 노드/유닛) : `y = w*x + b` 계산을 수행하는 최소 단위
   - 입력을 받아 자기만의 w, b로 계산한 뒤 활성화 함수(activation)를 거쳐 출력함
2. Layer (층) : 여러 Node를 묶은 단위
   - 이전 층의 출력이 다음 층의 모든 노드 입력으로 들어감
3. Sequential (모델) : Layer를 입력층 → 은닉층 → 출력층 순서로 일렬로 쌓은 모델

Node 여러 개가 Layer를 이루고, Layer 여러 개를 쌓으면 Sequential 모델이 됨

````python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    # 입력 특성 4개 -> 첫 은닉층, 노드(unit) 8개
    layers.Dense(units=8, activation='relu', input_shape=(4,)),

    # 두 번째 은닉층, 노드(unit) 4개
    layers.Dense(units=4, activation='relu'),

    # 출력층, 노드(unit) 1개 (예: 회귀 문제라 활성화 함수 없음)
    layers.Dense(units=1)
])

model.summary()
````

**코드의 인자**
1. `units` (=node 개수) : 그 층에 `y=wx+b` 계산이 몇 개 있는지
2. `activation` : 각 노드의 출력을 그대로 쓸지, 비선형으로 바꿔서 쓸지
3. `Sequential` : 층들을 입력 → 출력 방향으로 순서대로 연결한 것

## 7. Dense (완전연결층)

Fully Connected Layer. 이전 층의 모든 노드 출력이 다음 층의 모든 노드 입력으로 연결되는 층

**이름의 의미**
- 노드끼리 연결선이 빈틈없이 빽빽하게(dense) 차 있어서 Dense라고 부름

**사용법**
- `Dense(units=n)` : 이 층에 `y=wx+b` 계산(node)이 n개 있다는 뜻

## 8. 층 쌓기와 하이퍼파라미터

**층 쌓기 규칙**
1. `Dense`를 여러 번 `add`해서 층을 깊게 쌓을 수 있음
2. 첫 번째 층만 `input_dim`(입력 개수)을 지정함
   - 이후 층들은 이전 층의 출력과 자동으로 연결됨

```python
model.add(Dense(3, input_dim=1))  # 첫 층: 입력 개수 지정
model.add(Dense(5))               # 이후: 생략 가능, 자동으로 이어짐
model.add(Dense(4))
model.add(Dense(1))
```

### 8-1. 하이퍼파라미터

층 개수, 각 층의 노드(unit) 개수처럼 사람이 미리 정해야 하는 값

**가중치와의 차이**
- w, b는 학습 과정에서 자동으로 찾아지는 값
- 하이퍼파라미터는 학습 전에 사람이 정하는 값

**하이퍼파라미터 튜닝**
- 조합을 바꿔가며 loss가 더 잘 줄어드는 구조를 찾는 과정

## 9. 히든 레이어

입력층과 출력층 사이에 있는 중간 층

**층의 종류**
1. 인풋 레이어 : 노드 수가 x의 feature 개수로 정해짐 → 사람이 직접 넣는 값
2. 아웃풋 레이어 : 노드 수를 정답(y)에 맞춰 사람이 지정함 → 사람이 직접 다루는 값
3. 히든 레이어 : 그 사이의 중간 계산 결과
   - 사람이 값을 직접 정해줄 수도, 확인할 수도 없음

**이름의 의미**
- 인풋/아웃풋은 사람이 직접 다루는 값이라 겉으로 드러나 있음
- 히든 레이어의 출력은 모델이 스스로 계산해서 만든 중간 표현이라 사람 눈에는 숨겨져(hidden) 있음

### 9-1. 층·노드 개수 정하기

- 데이터(input)와 정답(output)은 사람이 넣어주지만, 그 사이를 몇 개의 층·노드로 채워야 최적인지는 정해진 답이 없음
- 여러 조합을 직접 시도하며 loss가 가장 잘 줄어드는 구조를 찾아야 함 (하이퍼파라미터 튜닝, §8-1)

## 10. epoch / batch_size / iteration

**용어**
1. epoch : 전체 데이터를 처음부터 끝까지 한 번 다 도는 것
2. batch_size : 가중치를 1번 업데이트할 때 사용하는 데이터 묶음 크기
3. iteration : 1 epoch 안에서 가중치가 업데이트되는 횟수
   - iteration = 전체 데이터 수 / batch_size

**예시 (데이터 9만 개)**
- 9만 개를 통으로 넣고 1 epoch → 가중치 업데이트 1번 (batch_size=9만)
- 3만 개씩 잘라서 1 epoch → 가중치 업데이트 3번 (batch_size=3만)
- 같은 데이터를 같은 만큼 봐도 후자가 가중치를 더 자주 고침

### 10-1. mini-batch의 장점

1. 업데이트 횟수 증가 : 방향을 자주 수정해 더 빠르고 정확하게 수렴함
2. 기울기의 노이즈 : 전체 데이터로 계산한 기울기는 정확하지만, 지역 최솟값·안장점에 갇히면 빠져나오지 못함
   - mini-batch는 매번 다른 부분집합이라 기울기에 노이즈가 생김
   - 그 노이즈 덕에 얕은 지역 최솟값에서 빠져나오기 쉬움
3. 메모리 : 순전파·역전파 때 각 층의 출력값을 모두 들고 있어야 함
   - 배치가 너무 크면 메모리 초과(OOM)가 날 수 있음

### 10-2. batch 크기의 트레이드오프

| batch 크기 | 기울기 | 메모리 | 업데이트 빈도 | 지역 최솟값 |
|---|---|---|---|---|
| **크다** | 안정적 | 많이 씀 | 적음 | 갇히기 쉬움 |
| **작다** | 노이즈 심함 (너무 작으면 불안정) | 적게 씀 | 자주 | 잘 안 갇힘 |

**사용 규칙**
1. 보통 2의 거듭제곱(32, 64, 128 ...)을 씀 → GPU 메모리 구조상 효율적
2. `model.fit(x, y, epochs=10, batch_size=32)` → batch_size를 생략하면 기본값 32
3. `batch_size=len(x)`로 주면 데이터를 통으로 학습하는 것과 같음
