### 3. AI 수학 기초

선형 관계식:

$$y = w \cdot x + b$$

(w: weight/가중치, b: bias/편향)

#### 학습 = "반복"이라는 것의 의미

1. w, b를 무작위로 초기화 → 당연히 예측이 틀림
2. 한 번 계산해서 **오차**를 구함
3. 오차를 줄이는 방향으로 w, b를 아주 조금씩 수정
4. **[예측 → 오차 계산 → w,b 조금 수정]** 을 수백~수천 번(epoch) 반복 → 오차가 점점 줄며 w, b가 정답에 근접

→ "반복"이 곧 학습의 실체: 한 번의 계산이 아니라 여러 번 고쳐나가는 과정 자체가 학습(training)

#### 최적의 weight = 최소의 오차

- "최적의 weight를 찾는다" == **"오차(loss)가 가장 작아지는 w, b 값을 찾는다"**
- 오차가 0에 가까울수록 예측이 실제값에 가까움 → 오차를 최소로 만드는 지점 = 최적의 weight

**경사 하강법(Gradient Descent)**: 최소 오차 지점을 찾아가는 방법
- 오차 함수의 기울기(경사)를 따라 내려가면서 w를 갱신
- 기울기가 0에 가까워지는 지점 = 오차가 최소가 되는 지점(최적의 weight)

- 딥러닝은 머신러닝의 한 종류, LLM은 딥러닝의 한 종류 (포함 관계, 별개 아님)
- LLM = Layer를 아주 깊고 크게(파라미터 방대) 쌓아 언어 데이터를 학습시킨 딥러닝 모델
- → 구조 원리는 동일, 규모(층 수/노드 수/데이터 양)만 압도적으로 커진 것

#### 공통 원리
머신러닝, 딥러닝 둘 다 본질은 동일:
1. 예측값을 만든다 (y = w*x + b 같은 함수)
2. 실제값과 예측값의 차이(오차, loss)를 계산한다
3. 오차를 줄이는 방향으로 w, b를 조금씩 수정한다 (경사 하강법)
4. 이 과정을 반복해서 최적의 w, b를 찾는다

#### 차이점 - 왜 딥러닝이 따로 불리는가

| 구분 | 구조 | 특징 |
|---|---|---|
| **머신러닝** (전통 방식) | `y = w*x + b` 계산을 단순한 구조로 1~소수 개만 사용 (선형회귀, 로지스틱회귀, 결정트리 등) | 사람이 어떤 feature를 볼지 미리 설계해야 하는 경우가 많음 |
| **딥러닝** (Deep Learning) | Node → Layer → 여러 Layer를 "깊게(Deep)" 쌓아올린 구조 | 층이 깊어질수록 더 복잡한 패턴을 스스로 찾아냄 |

**정리**: 딥러닝 = "여러 층(Layer)을 깊게 쌓은" 머신러닝<br/>
(경사 하강법으로 w, b를 찾는다는 학습 원리 자체는 동일)

### 5. 인공지능 코딩의 4단계

1. **데이터** 준비 (x, y)
2. **모델 구성** (Sequential + Dense)
3. **컴파일 & 훈련** (compile, fit)
4. **평가 & 예측** (evaluate, predict)

#### compile(loss, optimizer)
- **loss**: 오차를 측정하는 기준
  - `mse` (Mean Squared Error, 평균제곱오차): 오차를 **제곱**해서 평균 낸 값
  - `mae` (Mean Absolute Error, 평균절대오차): 오차의 **절댓값**을 평균 낸 값
- **optimizer**: loss를 최소화하는 방향으로 w, b를 실제로 갱신해주는 알고리즘 (경사 하강법 계열). 대표: `adam`, `sgd`, `rmsprop`
- → loss가 "얼마나 틀렸는지"를 재는 기준이라면, optimizer는 "그 오차를 보고 w, b를 어떻게 고칠지" 정하는 역할

#### fit / evaluate / predict
| 메서드 | 역할 |
|---|---|
| `model.fit(x, y, epochs=n)` | epoch 수만큼 [예측 → 오차 계산 → w,b 수정]을 반복하며 학습 |
| `model.evaluate(x, y)` | compile에서 지정한 loss 방식으로, 지금 모델 상태의 오차를 다시 계산 |
| `model.predict(x)` | 최종(마지막) 학습된 w, b를 대입해서 값을 계산 |

- epoch를 무조건 많이 돌린다고 좋은 건 아님 → 어느 지점부터 정체되거나 과적합 위험
- 데이터가 완벽한 비례 관계가 아니면(예: x=[1,2,3,4,5], y=[1,2,4,3,5]) 모든 점을 지나는 직선이 없어서 loss가 0으로 수렴하지 못하고, 오차가 가장 작은 "최적의 직선"에서 수렴함

### 6. Keras: Sequential / Layer / Node(Unit)

신경망 = `y = w*x + b` 계산을 여러 층(layer)으로 겹겹이 쌓고, 각 층마다 여러 번(node) 반복하는 구조

- **Node (=Unit, 노드/유닛)**: `y = w*x + b` 계산을 수행하는 최소 단위. 입력을 받아 자기만의 w, b로 계산 후 활성화 함수(activation)를 거쳐 출력
- **Layer (층)**: 여러 Node를 묶은 단위. 이전 층의 출력이 다음 층의 모든 노드 입력으로 들어감
- **Sequential (모델)**: Layer를 입력→은닉층→...→출력층 순서로 일렬로 쌓은 모델

→ 순서: **Node 여러 개 → Layer 구성 → Layer 여러 개가 쌓여 Sequential 모델** 완성

```python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    # 입력 특성 4개 -> 첫 은닉층, 노드(unit) 8개
    layers.Dense(units=8, activation='relu', input_shape=(4,)),

    # 두 번째 은닉층, 노드(unit) 4개
    layers.Dense(units=4, activation='relu'),

    # 출력층, 노드(unit) 1개 (예: 회귀 문제라 활성화 함수 없음)
    layers.Dense(units=1)
])

model.summary()
```

**정리**
- `units`(=node 개수): 그 층에 y=wx+b 계산기가 몇 개 있는지
- `activation`: 각 노드의 출력을 그대로 쓸지, 비선형으로 바꿔서 쓸지
- `Sequential`: 이 층들을 입력→출력 방향으로 순서대로 연결한 것

### 7. Dense 레이어로 모델 구성하기

**Dense** = 완전연결층(Fully Connected Layer)

- 이전 층의 모든 노드 출력이 다음 층의 모든 노드 입력으로 연결됨
- → 노드끼리 연결선이 빈틈없이 빽빽하게(dense) 꽉 차서 "Dense"라는 이름
- `Dense(units=n)` → 이 층에 y=wx+b 계산기(node)가 n개 있다는 뜻

### 8. 여러 층(Layer) 쌓기 & 하이퍼파라미터

- `Dense`를 여러 번 `add`해서 층을 깊게 쌓을 수 있음
- **첫 번째 층만** `input_dim`(입력 개수)을 지정해주면 됨 → 이후 층들은 자동으로 이전 층의 출력과 연결됨

```python
model.add(Dense(3, input_dim=1))  # 첫 층: 입력 개수 지정
model.add(Dense(5))               # 이후: 생략 가능, 자동으로 이어짐
model.add(Dense(4))
model.add(Dense(1))
```

#### 하이퍼파라미터란?
- 층 개수, 각 층의 노드(unit) 개수처럼 **사람이 미리 정해야 하는 값**
- w, b처럼 학습 과정에서 자동으로 찾아지는 값과는 다름
- 이 조합을 바꿔가며 loss가 더 잘 줄어드는 구조를 찾는 과정 = **하이퍼파라미터 튜닝**

### 9. 히든 레이어(Hidden Layer)가 "히든"인 이유

- **인풋 레이어**: 노드 수 = 사람이 데이터로 직접 넣어주는 값 (x의 feature 개수) → 눈에 보임
- **아웃풋 레이어**: 노드 수 = 사람이 정답(y)에 맞춰 직접 지정하는 값 → 눈에 보임
- **히든 레이어**: 그 사이의 중간 계산 결과 → 사람이 "이 값이어야 한다"고 직접 정해줄 수도, 확인할 수도 없는 값

→ 인풋/아웃풋은 사람이 직접 다루는 값이라 겉으로 드러나 있지만<br/>
히든 레이어의 출력값은 모델이 스스로 계산해서 만들어내는 중간 표현이라 사람 눈에는 "숨겨져(hidden)" 있음 → 그래서 **히든 레이어**

#### 그래서 층 개수/노드 개수는 사람이 "안다"기보다 "찾는다"
- 데이터(input)와 정답(output)은 사람이 넣어주지만, 그 사이를 몇 개의 층·노드로 채워야 최적인지는 아무도 정확히 모름
- → 정해진 정답이 없으니 여러 조합을 직접 시도해보며 loss가 가장 잘 줄어드는 구조를 찾아야 함 = **하이퍼파라미터 튜닝** (8번 섹션 참고)

### 10. epoch vs batch_size vs iteration

| 용어 | 의미 |
|---|---|
| **epoch** | 전체 데이터를 처음부터 끝까지 한 번 다 도는 것 |
| **batch_size** | 가중치를 1번 업데이트할 때 사용하는 데이터 묶음 크기 |
| **iteration** | 1epoch 안에서 가중치가 업데이트되는 횟수 = 전체 데이터 수 / batch_size |

예) 데이터 9만개
- 9만개 통으로 넣고 1epoch → 가중치 업데이트 **1번** (batch_size=9만, 기본값 안 씀)
- 3만개씩 잘라서 3epoch → 가중치 업데이트 **3번** (batch_size=3만)
- → 같은 데이터를 같은 만큼 봐도, 후자가 가중치를 더 자주 고쳐나감

#### mini-batch가 더 효율적인 이유
1. **업데이트 횟수 증가**: 조금씩 자주 걸으며 방향을 재확인 → 더 빠르고 정확하게 수렴
2. **노이즈 낀 기울기가 오히려 도움**: 전체 데이터로 계산한 기울기는 정확하지만 지역 최솟값/안장점에 갇히면 못 빠져나옴<br/>
→ mini-batch는 매번 다른 부분집합이라 기울기에 노이즈가 생기고, 이게 얕은 지역 최솟값을 흔들어 탈출을 도와줌
3. **메모리 문제**: 순전파/역전파 시 각 층의 출력값을 다 들고 있어야 함 → 배치가 너무 크면 메모리 초과(OOM) 위험

#### 트레이드오프
| batch 크기 | 기울기 | 메모리 | 업데이트 빈도 | 지역 최솟값 |
|---|---|---|---|---|
| **크다** | 안정적 | 많이 씀 | 적음 | 갇히기 쉬움 |
| **작다** | 노이즈 심함 (너무 작으면 불안정) | 적게 씀 | 자주 | 잘 안 갇힘 |

- 실무 관행: 2의 거듭제곱(32, 64, 128 ...) - GPU 메모리 구조상 효율적
- `model.fit(x, y, epochs=10, batch_size=32)` - batch_size 생략 시 기본값 32<br/>
`batch_size=len(x)` → 지금까지처럼 데이터를 통으로 학습하는 것과 동일
