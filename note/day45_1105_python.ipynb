## 1. 모듈

함수나 변수를 모아 둔 `.py` 파일. 다른 파일에서 불러 씀

같은 함수를 파일마다 다시 쓰면 고칠 때 전부 손봐야 함. 한 파일에 두고 불러 쓰면 고칠 자리가 한 군데

1. `import p01_car` : 파일 전체를 불러옴. 쓸 때는 `p01_car.drive()`
2. `from p01_car import drive` : 함수만 불러옴. 쓸 때는 `drive()`

- 모듈 이름은 파일 이름에서 `.py`를 뗀 것
- 같은 폴더에 있으면 경로를 적지 않아도 찾음. 파이썬이 실행하는 스크립트가 있는 폴더를 먼저 뒤지기 때문
- `import`는 쓸 때마다 모듈 이름을 앞에 붙여서 어느 파일에서 온 함수인지 드러남
- `from`은 이름을 바로 써서 짧은 대신, 다른 모듈에 같은 이름이 있으면 나중에 불러온 것이 앞의 것을 가림
- 처음 불러올 때 `__pycache__` 아래에 `.pyc`가 생김. 다음부터는 파일을 다시 해석하지 않고 이것을 씀

## 2. __name__

파이썬이 파일마다 자동으로 넣어 두는 모듈 이름 변수

`import`는 그 파일을 위에서 아래로 한 번 실행하므로, 파일 안에 적어 둔 함수 호출도 같이 돌아감. 직접 실행된 것인지 불려 온 것인지 구분해야 그것을 막을 수 있음

1. 직접 실행하면 `__name__`에 `__main__`이 들어감
2. `import`되면 모듈 이름(`p11_car`)이 들어감

- 앞뒤의 밑줄 두 개는 파이썬이 쓰는 이름이라는 표시. 내가 만들어 넣는 변수가 아님
- 불러오기만 하려는데 호출까지 돌아가면 쓰지도 않을 출력이 섞임
- 돌리는 파일이 바뀌면 같은 파일의 `__name__`도 바뀜. 어느 쪽에서 돌렸는지에 따라 값이 달라짐

## 3. __main__ 블록

그 파일을 직접 실행할 때만 돌리는 블록

함수를 만들고 바로 불러 확인하는 줄을 그냥 두면 `import`할 때도 돌아감. 그 줄만 `if` 안으로 넣으면 불러오는 쪽에서는 건너뜀

1. `if __name__=='__main__':` : 아래에 직접 돌릴 때만 할 일을 둠
2. `def` : `if` 바깥에 둬서 불러오는 쪽에서 쓸 수 있게 함

```python
def drive():
    print("운전하다")

if __name__=='__main__':
    drive()
```

- **다른 파일에서 쓸 것은 `if` 바깥에, 직접 돌려 확인할 것은 `if` 안에 둠**
- 불러오는 쪽에서는 `if` 안이 돌지 않고 함수 정의만 들어옴
- 모듈을 만들 때 테스트 호출을 지우지 않고도 남겨 둘 수 있음

## 4. 모듈별 이름공간

전역 변수는 그 변수를 적어 둔 모듈 안에서만 전역

여러 파일에 같은 이름의 변수가 있어도 서로 섞이지 않아야 함. 어느 쪽 값을 쓰는지는 변수가 어디에 적혀 있는지로 정해짐

```python
# p31_sample.py
x = 111

def test():
    print('x: ', x)
```

1. 불러오는 쪽에도 `x`를 따로 두고 값을 다르게 적음
2. 불러오는 쪽 함수는 그쪽 `x`를 찍고, `p31_sample.test()`는 `p31_sample`의 `x`를 찍음

- **함수가 보는 변수는 어디서 불렀는지가 아니라 어디에 정의됐는지로 정해짐**
- `from p31_sample import test`로 함수만 가져와도 결과가 같음. 이름만 옮겨 온 것이고 함수가 보는 전역은 원래 모듈 쪽
- 그래서 불러온 모듈의 변수를 바꾸려면 그 모듈 이름을 붙여 `p31_sample.x`로 적어야 함

## 5. 패키지

모듈을 폴더로 묶어 점으로 내려가며 불러 쓰는 것

파일이 늘면 한 폴더에 다 두기 어렵고 이름도 겹침. 폴더로 나누면 `machine.car`처럼 경로가 이름에 들어가서 구분됨

1. `import machine.car as car` : 폴더.모듈까지 적고 짧은 이름을 붙임
2. `from machine.car import drive` : 폴더.모듈에서 함수만 꺼냄
3. `from machine import car, tv` : 폴더에서 모듈 여러 개를 한 줄로 꺼냄
4. `from machine.test import car, tv` : 폴더 안의 폴더도 점으로 이어서 내려감

- 점의 왼쪽은 폴더, 맨 오른쪽은 모듈이나 그 안의 이름
- `import machine.car`만 쓰면 부를 때도 `machine.car.drive()`로 다 적어야 해서 `as`로 줄임
- `from machine import car`는 모듈을 가져와서 `car.drive()`, `from machine.car import drive`는 함수를 가져와서 `drive()`
- `machine/car.py`와 `machine/test/car.py`처럼 이름이 같아도 경로가 달라 섞이지 않음
- `__init__.py`를 두지 않아도 불러짐. 3.3부터 폴더만 있어도 패키지로 인식함
