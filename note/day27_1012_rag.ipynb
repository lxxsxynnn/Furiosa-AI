## 1. 대화 기록

이전 대화를 프롬프트에 넣어 대화를 이어 가는 방법 (`rag21_Gradio_history`)

모델은 `invoke`마다 그 프롬프트만 받음. 이전 대화가 없으면 "아까 말한 것" 같은 질문에 답할 수 없음

1. `MessagesPlaceholder(variable_name='chat_history')` : 메시지 리스트가 들어갈 자리. `system`과 `human` 사이에 둠
2. 메시지 변환 : `history`를 `role`에 따라 `HumanMessage`·`AIMessage`로 바꿈
3. `invoke({"input": ..., "chat_history": ...})` : 자리 이름과 같은 키로 넘김

- Gradio 6 `history`는 `role`·`content` dict 리스트. 이번 질문은 없음
- `content`는 `[{'type': 'text', 'text': ...}]` 형식. 메시지 클래스에 그대로 넣어도 됨
- `create_retrieval_chain`은 `input` 외의 키도 프롬프트까지 넘김
