## 1. 검색 문서 확인

체인 결과의 `context`로 답에 쓰인 청크를 확인하는 방법 (`rag22_Gradio_1_doc`)

답만 보면 검색이 맞았는지 알 수 없어, 프롬프트에 들어간 청크를 터미널에 출력함

1. `response['context']` : 검색된 `Document` 리스트
2. `doc.metadata.get("source")` : 청크가 나온 파일 경로
3. `doc.page_content` : 청크 원문

- `metadata.get`은 키가 없으면 에러 대신 `None`을 돌려줌

## 2. 유사도 기준값

유사도 점수가 기준보다 낮은 청크를 검색 결과에서 빼는 방식 (`rag22_Gradio_2_similarity`)

기본 검색기는 질문과 상관없어도 `k`개를 돌려줘서, 관련 없는 청크가 프롬프트에 들어감

1. `search_type="similarity_score_threshold"` : 기준값 검색으로 바꿈
2. `score_threshold` : 0~1 유사도 점수 기준. 이 값 이상만 남김
3. `k` : 기준을 넘은 청크 중 최대 개수

- 기준을 넘은 청크가 없으면 `No relevant docs were retrieved ...` 경고가 나오고 `context`가 빈 리스트
- `context`가 비면 "답변할 수 없다"는 지시에 따라 모델이 인사 같은 일상 대화에도 같은 문장으로 답함

## 3. 프롬프트 규칙

system 프롬프트에 질문 종류별 답하는 방법을 번호로 적는 방법 (`rag22_Gradio_3_prompt`)

기준값 검색만 쓰면 일상 대화에도 "답변할 수 없습니다"가 나와서, 문서 질문과 다른 질문을 나눠 지시함

1. 이전 대화 : 이전 대화에서 확인되는 정보는 대화 기록을 참고
2. 일상 대화 : 인사·잡담은 컨텍스트 없이 답함
3. 문서 질문 : 컨텍스트를 근거로 답함
4. 정보 없음 : 문서 질문인데 컨텍스트에 정보가 없으면 "답변할 수 없다"고 답함

- 규칙은 프롬프트에만 들어가고, 검색 방식은 §2와 같음
