## 1. 활성화 함수 그래프

활성화 함수를 식으로 직접 만들어 x 구간에 대입해 모양을 보는 것 (`keras78_01_simoid` ~ `keras78_08_leaky_lelu`)

`activation='relu'`처럼 이름만 넣어 쓰면 값이 어떻게 바뀌는지 보이지 않음. 식을 그려 보면 입력에 따라 출력이 어디까지 가는지, 어느 구간에서 기울기가 0이 되는지 보임

| 함수 | 식 | 입력이 음수면 | 입력이 양수면 |
|---|---|---|---|
| sigmoid | `1 / (1+ np.exp(-x))` | 0으로 수렴 | 1로 수렴 |
| tanh | `np.tanh(x)` | -1로 수렴 | 1로 수렴 |
| relu | `np.maximum(0, x)` | 0 | 입력과 같음 |
| silu | `x * (1 / (1 + np.exp(-x)))` | 0에 가까운 작은 음수 | 입력과 거의 같음 |
| mish | `x * np.tanh(np.log(1 + np.exp(x)))` | 0에 가까운 작은 음수 | 입력과 거의 같음 |
| elu | `(x > 0) * x + (x < 0) * (alpha * (np.exp(x) - 1))` | `-alpha`로 수렴 | 입력과 같음 |
| selu | `lmbda * ((x > 0) * x + (x <= 0) * (alpha * (np.exp(x) - 1)))` | `-alpha * lmbda`로 수렴 | 입력의 `lmbda`배 |
| leaky_relu | `np.where(x > 0, x, alpha * x)` | 입력의 `alpha`배 | 입력과 같음 |

1. `np.arange(-5, 5, 0.1)`과 `plt.plot(x, y)` : x 구간을 만들어 함수에 넣고 그림
2. `alpha` : 음수 입력을 얼마나 통과시킬지 정하는 값. leaky_relu는 0.01, elu는 1을 씀
3. `lmbda` : selu에서 출력 전체에 곱하는 값. `alpha`와 함께 논문이 제안한 고정값

- sigmoid와 tanh는 입력이 커져도 출력이 더 변하지 않아서 그 구간의 기울기가 0에 가까워짐
- relu는 양수 구간에서 출력이 입력과 같아 기울기가 1로 일정함. 층을 쌓아도 기울기가 줄지 않음
- relu는 입력이 음수면 출력도 기울기도 0이라 그 노드의 가중치가 갱신되지 않음
- leaky_relu와 elu는 음수 쪽에 기울기를 조금 남겨서 갱신이 멈추는 구간을 줄임
- silu와 mish는 0 근처가 곡선이라 relu처럼 각지지 않고 음수 쪽도 조금 통과시킴
- mish는 silu와 모양이 거의 같은데 `exp`·`log`·`tanh`를 차례로 계산해서 연산량이 더 많음
- silu는 2017년 구글이 swish라는 이름으로 발표함

#### cf) 넘파이로 식 적기

- `np.maximum(0, x)`은 두 값을 비교해 큰 쪽을 돌려줌. 배열 전체에 한 번에 적용됨
- `(x > 0) * x`는 조건이 참이면 1, 거짓이면 0으로 곱해지는 것을 이용해 구간을 나눈 것
- `np.where(조건, 참일 때, 거짓일 때)`는 구간을 곱셈 없이 그대로 적음. `np.maximum(alpha * x, x)`로도 같은 모양이 나옴
- **`lambda`는 파이썬 예약어라 변수 이름으로 쓸 수 없어서 `lmbda`로 적음**
- selu의 `alpha`·`lmbda`는 논문이 제안한 고정값이고, elu의 `alpha`는 넣어 보는 값
