## 1. map

함수를 리스트의 값마다 한 번씩 적용해 주는 것 (`p01_map`)

for문으로 돌면서 새 리스트에 담는 코드를 한 줄로 줄임

1. `map(multiply, numbers)` : 앞은 적용할 함수 이름, 뒤는 값이 든 반복 가능한 객체
2. `list(result)` : map 객체를 리스트로 바꿔서 값을 꺼냄

- 함수는 이름만 넘김. `multiply(x)`처럼 괄호를 붙이면 호출한 결과가 넘어감
- 그대로 `print`하면 `<map object ...>`만 나옴. 꺼내기 전에는 값을 만들어 두지 않기 때문
- map 객체는 한 번 꺼내면 비어서, `list()`를 다시 불러도 빈 리스트가 나옴

## 2. lambda

이름 없이 한 줄로 만드는 함수 (`p02_lambda1` ~ `p02_lambda3`)

`map`에 넘길 함수처럼 한 번만 쓰는 함수는 `def`로 따로 만들면 이름과 줄만 늘어남

1. `aaa = lambda x : x * 2` : 콜론 왼쪽이 매개변수, 오른쪽이 돌려줄 값
2. `lambda x, y: x + y` : 매개변수가 여러 개면 콤마로 적음
3. `lambda x: "합격" if x >= 60 else "불합격"` : 조건식도 한 줄로 넣음
4. `list(map(lambda x: "합격" if x >= 60 else "불합격", scores))` : `map`에 바로 넘겨 씀

- `return`을 적지 않음. 오른쪽 식의 결과가 그대로 반환값임
- `def multiply(x): return x * 2`와 `lambda x: x * 2`는 같은 일을 함
- 식 하나만 들어감. if문·for문처럼 여러 줄이 필요하면 `def`로 만듦
- 조건식은 `값 if 조건 else 값` 순서라서 참일 때 값을 앞에 씀
- 변수에 넣어 두면 `aaa(10)`처럼 이름으로 부를 수 있음
