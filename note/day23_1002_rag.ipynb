# 1002 23일차

## 1. 의미 검색

질문과 뜻이 가까운 문서를 벡터로 찾는 검색. 벡터 검색이라고도 함

| | 키워드 검색 | 의미 검색 |
|---|---|---|
| 기준 | 단어가 들어 있는지 | 뜻이 비슷한지 |
| "강아지 사료" | "강아지", "사료"가 있는 문서 | "반려견 먹이"까지 |

1. 문서를 임베딩해 저장
2. 질문도 같은 모델로 임베딩
3. 코사인 유사도로 가까운 문서를 찾음 (day22 RAG §1-2)

### 1-1. 벡터 저장소

| | Chroma | FAISS |
|---|---|---|
| 종류 | 문서형 벡터 DB | 벡터 검색 라이브러리 (메타) |
| 기본 단위 | 컬렉션 | 인덱스 |
| 저장 | 벡터 + 원문 + 메타데이터 | 벡터 |
| 핵심 | 문서 관리 | 대규모 검색 속도 |

1. 컬렉션 : 문서를 묶어 관리하는 단위
2. 인덱스 : 벡터를 담아 검색하는 단위. 종류를 골라 성능을 튜닝함

- Chroma : 질문 벡터 > 컬렉션 검색 > 문서 + 메타데이터 반환
- LangChain은 둘 다 `VectorStore`로 래핑해서 같은 방법으로 씀
