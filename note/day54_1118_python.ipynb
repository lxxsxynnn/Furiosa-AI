## 1. 데코레이터

함수를 받아 기능을 덧붙인 함수를 돌려주는 함수 (`p03_Decorator`)

여러 함수에 같은 코드를 앞뒤로 넣어야 할 때, 함수마다 적지 않고 한 번 만들어 붙임

1. `def deco(func):` : 꾸밀 함수를 매개변수로 받음
2. `def wrapper():` : 원래 함수 호출을 앞뒤 동작으로 감싼 함수를 안에 만듦
3. `return wrapper` : 괄호 없이 함수 자체를 돌려줌
4. `@deco` : 꾸밀 함수 정의 위에 붙임. `hello2 = deco(hello2)`와 같음

- **원래 함수의 코드를 고치지 않고 동작을 덧붙임**
- `func()`를 적은 자리에서 원래 함수가 실행되므로, 그 위아래가 실행 전·후가 됨
- `return wrapper()`로 쓰면 함수가 아니라 그 자리에서 실행한 결과가 돌아감
- `@deco`를 붙인 뒤 `hello2()`를 부르면 실제로는 `wrapper()`가 돌아감
- 본래 기능과 상관없는 일(실행 시간 측정·로그 기록·권한 검사·예외 처리·호출 횟수)에 씀

## 2. 실행 시간 측정 데코레이터

함수가 도는 데 걸린 시간을 재서 찍어 주는 데코레이터 (`p04_Deco_timer`)

함수마다 시각을 재는 줄을 넣으면 재는 코드가 본래 기능 사이에 섞임. 데코레이터로 빼면 함수에는 할 일만 남음

1. `start = time.perf_counter()` : `func()` 앞에서 시각을 잼
2. `func()` : 원래 함수를 실행함
3. `end = time.perf_counter()` : 끝난 시각을 잼
4. `round(end - start, 6)` : 초 단위 차이를 소수점에서 잘라 찍음

- **재는 코드는 데코레이터에만 있고, 꾸밀 함수에는 `@timer` 한 줄만 붙임**
- 같은 데코레이터를 여러 함수에 붙여도 각 함수의 시간이 따로 나옴
- `time.sleep(1)`을 넣은 함수는 1보다 조금 크게 나옴. 함수 호출과 출력에 드는 시간이 같이 들어감
- `time.time()`은 시계 시각이라 짧은 구간을 재는 데는 `perf_counter()`를 씀
