## 1. split_x

수열을 `size` 크기로 잘라 쌓는 함수. 시계열에서 x와 y를 만들 때 씀

```python
def split_x(dataset, size):
    arr = []
    for i in range(len(dataset) - size + 1):
        subset = dataset[i : (i + size)]
        arr.append(subset)
    return np.array(arr)
```

### 1-1. 동작

`a = [1 ... 10]`, `size=5`일 때

```
i=0   a[0:5]    [1 2 3 4 5]
i=1   a[1:6]    [2 3 4 5 6]
      ...
i=5   a[5:10]   [6 7 8 9 10]
```

- 한 칸씩 밀며 떼어내 쌓은 뒤 `np.array`로 묶음
- 조각끼리 값이 겹침
- 축이 하나 늘어남. `(10,)` → `(6, 5)`, `(10, 2)` → `(6, 5, 2)`

### 1-2. 행 개수

| 하는 일 | 행 개수 | 예 (길이 10) |
|---|---|---|
| 자르기만 | `len - size + 1` | `size=5` → 6행 |
| x와 y까지 나눔 | `len - timesteps` | `timesteps=3` → 7행 |

- `+1`은 `i`가 `0`부터 `len - size`까지 가기 때문
- 아래 줄은 `size = timesteps + 1`로 자른 것과 같음 (day18 §4-1)

### 2-1. 대괄호를 이어 붙일 때

`(4, 3, 2)` 배열에서

| 쓰는 법 | 모양 | 꺼내는 것 |
|---|---|---|
| `arr[:, -1]` | `(4, 2)` | 묶음마다 마지막 줄 |
| `arr[:][-1]` | `(3, 2)` | 마지막 묶음 전체 |

- `arr[:][-1]`은 0번 축에 두 번 접근한 것이라 `arr[-1]`과 같음
