## 1. 케라스 모델 하이퍼파라미터 튜닝

케라스 모델을 sklearn 탐색기에 넣어 파라미터 조합을 찾는 것 (`keras77_hyperParameter02_diabetes`)

for문으로 돌리면 조합마다 반복문을 겹쳐 쓰고 교차 검증도 직접 나눠야 함. 조합 뽑기와 교차 검증을 sklearn의 `RandomizedSearchCV`에 넘김

1. `def build_model(drop=0.5, optimizer='adam', activation='relu', node1=128, ...)` : 모델을 함수로 만들고 바꿀 값은 매개변수로 받음
2. `def create_hyperparameter():` : 탐색할 값 목록을 딕셔너리로 모아 반환함
3. `KerasRegressor(build_fn=build_model, verbose=1)` : 함수를 sklearn이 쓰는 모델 모양으로 감쌈
4. `RandomizedSearchCV(keras_model, hyperparameters, cv=3, random_state=333, n_iter=10, refit=True)` : 감싼 모델을 탐색기에 넣음
5. `model.fit(x_train, y_train, epochs=100, callbacks=[es, rlr], validation_split=0.1)` : 케라스 쪽 인자는 `fit`에 그대로 넘김

- 래핑하기 전의 `build_model`은 함수라서 sklearn이 모델로 인식하지 못함
- **딕셔너리 키는 `build_model`의 매개변수 이름과 같아야 함**
- `batch_size`는 `build_model`에 없지만 래퍼가 받아서 `fit`에 넘김
- 회귀는 `KerasRegressor`, 분류는 `KerasClassifier`를 `tensorflow.keras.wrappers.scikit_learn`에서 가져옴
- `cv=3`에 `n_iter=10`이면 30번 학습하고, `refit=True`라서 마지막에 한 번 더 학습함
- `epochs`·`callbacks`·`validation_split`은 탐색기를 만들 때가 아니라 `fit`에 넣음
