## 1. RandomizedSearchCV

파라미터 조합 전체가 아니라 그중 일부만 무작위로 뽑아 교차 검증하는 탐색 (`m09_RandomSearchCV_01_digits`, `m09_RandomSearchCV_02_diabetes`)

`GridSearchCV`는 조합 수 × fold 수만큼 다 훈련해서 후보 값을 늘리는 만큼 시간이 늘어남. 비슷한 점수가 나오는 조합이 많아 전부 돌리지 않아도 좋은 조합을 찾음

1. `RandomizedSearchCV(xgb, parameters, cv=kfold, verbose=1, n_jobs=1, random_state=333, n_iter=10, refit=False)` : `GridSearchCV` 자리에 그대로 바꿔 씀
2. `n_iter=10` : 뽑을 조합 수. 기본값이 10
3. `random_state=333` : 어떤 조합이 뽑힐지 고정
4. `refit=False` : 최적 조합으로 다시 학습하는 마지막 1회를 건너뜀

- 전체 조합이 70개여도 `n_iter`가 10이면 10 × fold 수만 훈련함
- 뽑기 대상은 딕셔너리 목록 전체. 어느 딕셔너리에서 몇 개가 뽑힐지는 정해져 있지 않음
- `random_state`를 주지 않으면 돌릴 때마다 다른 조합이 뽑혀서 앞 결과와 견줄 수 없음
- `GridSearchCV`는 조합을 전부 돌려서 뽑을 때 쓸 난수가 없어 `random_state`를 두지 않음
- 조합을 다 보지 않으므로 찾은 조합이 전체 최적이라는 보장은 없음
- 결과를 꺼내는 자리(`best_params_`·`best_score_`)는 `GridSearchCV`와 같음

#### cf) refit

- `refit=True`(기본)는 탐색이 끝난 뒤 최적 조합으로 train 전체를 한 번 더 학습해서 `best_estimator_`에 넣음
- `model.score`·`model.predict`는 그 `best_estimator_`에 넘겨 주는 통로임. 탐색기 자체는 모델을 감싼 것이라 스스로 예측하지 않음
- **`refit=False`면 `best_estimator_`가 만들어지지 않아서 `model.score`·`model.predict`를 쓸 수 없음**
- `best_score_`·`best_params_`·`best_index_`는 교차 검증 결과표에서 뽑는 값이라 `refit=False`에서도 남음
- test 점수까지 보려면 `best_params_`로 모델을 다시 만들어 직접 학습함
- 조합만 찾아 두고 최종 모델은 따로 만들 때 `refit=False`로 둠
